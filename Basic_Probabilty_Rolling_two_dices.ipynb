```python
#Basic_Probability1: Rolling two dices
import random
from fractions import Fraction
total_trials=1000 #taking the random no. of trials
sum=0
for i in range(total_trials):
    die1=random.randint(1,6)

    die2=random.randint(1,6)

    total=die1+die2
    if total==7:
        sum+=1
prob_sum_7=Fraction(sum,total_trials)

print(f"The probability of getting 7 in total trials=={prob_sum_7}")

```
