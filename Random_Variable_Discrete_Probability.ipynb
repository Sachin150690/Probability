```python
import numpy as np

#Values
values=[1,2,3]

probabilities=[0.25,0.35,0.4]

#Generating the sample of size 1000
sample=np.random.choice(values,size=1000,p=probabilities)

print(sample)

#Mean Calculation
mean=np.mean(sample)

#Variance Calculation
var=np.var(sample)

#Standard Deviation
std=np.std(sample)

#Display Results
print(f"Mean of the samples = {mean}")
print(f"Variance of the sample {var}")
print(f"Standard Deviation {std}")
```
