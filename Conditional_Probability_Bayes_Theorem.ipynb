```python
#PartA:Conditional Probability
import random
from fractions import Fraction

#Bag Creation
bag=["Red"]*5+["Green"]*7+["Blue"]*9
print(bag)

No_of_draws=1000

draws=[]

for i in range(No_of_draws):
    ball=random.choice(bag)
    draws.append(ball)

print(draws)

previous_blue=0
current_red=0
for i in range(1,No_of_draws):
    if draws[i-1]=="Blue":
        previous_blue+=1

        if draws[i]=="Red":
            current_red+=1
#Conditional_Probablity of Red ball provided previous ball is blue
P_red_given_blue=current_red/previous_blue

print(f"No.of times previous ball is blue:{previous_blue}")
print(f"No.of times current ball is red:{current_red}")
print(f"P(Red | Previous Blue = {P_red_given_blue:.5f}")

#Part B: Verify using Bayes Theorem
red_count=draws.count("Red")
p_red=red_count/No_of_draws

blue_count=draws.count("Blue")
p_blue=blue_count/No_of_draws

blue_given_red=0
red_count_for_condition=0

for i in range(1,No_of_draws):
    if draws[i]=="Red":
        red_count_for_condition+=1

        if draws[i-1]=="Blue":
            blue_given_red+=1

p_blue_given_red=blue_given_red/red_count_for_condition

#now Applying Bayes Formula
p_red_given_blue=(p_blue_given_red*p_red)/p_blue

print(f"----Bayes Theorem Validation----")
print(f"P (Red |Blue) (Bayes Theorem) = {p_red_given_blue:.5f}")
```
