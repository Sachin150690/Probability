```python
import numpy as np
import matplotlib.pyplot as plt

n_samples=2000

exp_mean=5

#Generating Random Samples
samples=np.random.exponential(exp_mean,n_samples)
print(samples)

#Histogram

plt.figure()
plt.title("Histogram representing the samples")
plt.hist(samples,bins=30,edgecolor="black",density=True,label="Histogram")
plt.legend()
plt.grid()
plt.show()

#Probabilty Density Function Chart(PDF Chart)
x=np.linspace(0,max(samples),500)
#print(x)

#pdf calculation
pdf=(1/exp_mean)*np.exp(-x/exp_mean)
plt.figure()
plt.plot(x,pdf,linewidth=3.5,label="pdf")
plt.xlabel("Values")
plt.ylabel("Density")
plt.title("PDF Chart")
plt.legend()
plt.grid()
plt.show()
```
