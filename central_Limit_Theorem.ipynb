```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

#generating 10000 random numbers
random_numbers=np.random.uniform(0,1,10000)
print(random_numbers)

#Drawing 1000 samples of size 30
samples=np.random.choice(random_numbers,size=(1000,30))

print(samples)

#Calculating Sample Means
means=[]
for i in samples:
    means.append(np.mean(i))

print("**********************************************")
print(list(means))

#Plotting Uniform Distribution
plt.figure()
plt.title("Population Uniform Distribution")
plt.hist(random_numbers,bins=30,density=True,edgecolor="black")
plt.grid()
plt.show()

#Plotting Means of the Samples
plt.figure()
plt.title("Mean Distribution")
sns.histplot(means,kde=True)
plt.show()
```
