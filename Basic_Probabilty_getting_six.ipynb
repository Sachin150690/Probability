```python
from fractions import Fraction
import random

def Rolling_Dice(total_trials):
    trials=10
    sum=0
    for i in range(total_trials):

        for i in range(trials):
            choice=random.randint(1,6)
            if choice==6:
                sum+=1
                break
    return Fraction(sum,total_trials)

trials_input=int(input("Enter the total number of trials"))

print(f"The probability of getting at least one 6 is {Rolling_Dice(trials_input)}")
```
