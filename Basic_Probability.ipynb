```python
#Basic Probability1: Tossing Coin 1000 times
import random
from fractions import Fraction
total_trials=10000

heads=0
tails=0
for i in range(total_trials):
    result=random.choice(["Heads","Tails"])
    if result=="Heads":
        heads+=1
    else:
        tails+=1

prob_head=Fraction(heads,total_trials)
prob_tails=Fraction(tails,total_trials)

print(f"The probability of head= {prob_head}")
print(f"The probability of tails= {prob_tails}")
```
